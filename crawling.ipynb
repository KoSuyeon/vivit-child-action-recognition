```python
from pytube import YouTube
```

```python
# 다운로드할 유튜브 영상 주소 리스트
video_urls = [
    "url1",
    "url2",
    "..."
]
```

```python
# 영상 다운로드 함수
def download_video(url, output_path):
    try:
        yt = YouTube(url)
        video = yt.streams.get_highest_resolution()  # 가장 높은 해상도의 영상 스트림 선택
        video.download(output_path)
        print(f"{yt.title} 다운로드 완료")
    except Exception as e:
        print(f"다운로드 중 에러 발생: {e}")
```

```python
label = 'Clapping'
# 각 영상 주소에 대해 다운로드 수행
for url in video_urls:
    download_video(url, f'rawdata/{label}')  # 다운로드 받을 폴더 경로를 지정하세요
```
